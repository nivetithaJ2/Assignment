```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('/content/electricity_data.csv')

print(data.head())
print(data.shape)
```

```python
print(data.info())

print(data.describe())

print(data.isnull().sum())
```

```python
data = data.drop_duplicates()

data = data.dropna()

data['Date'] = pd.to_datetime(data['Date'])

data['Hour'] = pd.to_datetime(
    data['Time'], format='%H:%M'
).dt.hour

data['Month'] = data['Date'].dt.month
```

```python
print(data[['Temperature',
            'Electricity_Usage']].corr())
```

```python
sns.scatterplot(
    data=data,
    x='Temperature',
    y='Electricity_Usage'
)

plt.title('Temperature vs Electricity Usage')
plt.xlabel('Temperature')
plt.ylabel('Electricity Usage (kWh)')

plt.show()
```

```python
sns.heatmap(
    data[['Temperature',
          'Electricity_Usage']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Electricity Usage Correlation')

plt.show()
```

```python
daily_usage = data.groupby('Date')['Electricity_Usage'].sum()

print(daily_usage)
```

```python
monthly_usage = data.groupby('Month')['Electricity_Usage'].sum()

print(monthly_usage)
```

```python
hourly_usage = data.groupby('Hour')['Electricity_Usage'].mean()

print(hourly_usage)
```

```python
daily_usage.plot(kind='line', marker='o')

plt.title('Daily Electricity Consumption')
plt.xlabel('Date')
plt.ylabel('Electricity Usage (kWh)')

plt.show()
```

```python
monthly_usage.plot(kind='bar')

plt.title('Monthly Electricity Consumption')
plt.xlabel('Month')
plt.ylabel('Electricity Usage (kWh)')

plt.show()
```

```python
hourly_usage.plot(kind='line', marker='o')

plt.title('Average Electricity Usage by Hour')
plt.xlabel('Hour')
plt.ylabel('Average Electricity Usage (kWh)')

plt.show()
```
